# **Reading the raw EM-DATA Environmental Disaster Database**

```r
disaster_raw <- read_csv("raw_data/emdata_disaster_database.csv")
head(disaster_raw)
```

```r
#checking the names of all columns
colnames(disaster_raw)
```

```r
# Rename and select the columns of interest, keeping only disasters in USA and dropping disasters that are not meteorological, climatological, geophysical and or hydrological
disaster_clean <- disaster_raw %>%
  rename(
    year = `Start Year`,
    country = Country,
    disaster_group = `Disaster Group`,
    disaster_subgroup = `Disaster Subgroup`,
    disaster_type = `Disaster Type`,
    disaster_subtype = `Disaster Subtype`,
    total_USD_damage = `Total Damage, Adjusted ('000 US$)`
  ) %>%
  select(
    year,
    country,
    location = Location,
    disaster_group,
    disaster_subgroup,
    disaster_type,
    total_USD_damage
  ) %>%
  filter(
    country == "United States of America",
    disaster_subgroup != "Biological",
    !disaster_group %in% c(
      "Industrial Accident",
      "Miscellaneous Accident",
      "Transport"
    )
  )

head(disaster_clean)
```

```r
write_csv(disaster_clean, "clean_data/emdata_disaster_clean.csv")

# Check: the file should now be in clean_data/
list.files("clean_data")
```
